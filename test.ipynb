```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('/home/jingo-green/Desktop/imax/data/dataset_phishing_trimmed.csv')
df.head()
X = df.drop(columns=['status'])
y = df['status']
```

```python
X.head()
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, train_size=0.8, random_state=42, stratify=y)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
y_train.head()
```

```python
y_test.head()
```

```python
rf = RandomForestClassifier(random_state=42)
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
from datetime import datetime
samples = len(X_train)

for sub in range(5, 101, 5):
    start = datetime.now()
    subset = int(samples*(sub/100))
    subx = X_train.iloc[:subset] 
    suby = y_train.iloc[:subset]
    rfc = RandomForestClassifier(random_state=42)
    rfc.fit(subx, suby)
    y_pred = rfc.predict(X_test)
    end = datetime.now()
    print(f"duration={end-start}, subset={sub}%\n{classification_report(y_test, y_pred)}")
```
